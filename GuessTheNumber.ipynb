```python
import random
games_played = 0 #keeping track of the player's stats throughout the game
games_won = 0
games_lost = 0
```

```python
def get_guess(): 
    #asking the player for a valid guess between 1 and 100
    while True:
        try:
            guess = int(input("Enter your guess: "))
            #check if the input is between 1 and 100
            if 1<= guess <=100:
               return guess
            else:
                print( "Please enter a number between 1 and 100.")

        except ValueError:
            #print an error message if the player enters something that can't be converted into a number
            print("Invalid input. Please enter a number.")
```

```python
def play_game():
    #round one of the game
    global games_played, games_won, games_lost
    #add 1 to the number of games played
    games_played += 1 
    
    #generates a secret number (1 -100)
    secret_number = random.randint(1, 100)
    
    #gives the player a set number of attempts
    attempts = 3

    print("\n-------------------------------------------")
    print("         GUESS THE NUMBER GAME               ")
    print("---------------------------------------------")
    print("I'm thinking of a number between 1 and 100.")
    print("You have 3 attempts to guess it!")
    print("Good Luck!!!")

    #keep asking the player to guess while they have attempts remaining
    while attempts >0:
        
        #asks to input guesses and compare the guess with the secret number
        guess = get_guess()
        #check if the player's input is correct
        if guess == secret_number:
            print("\nWell Done! You have guessed the number!")
            print("Congratulations!!!")
            #Add 1 to the games won
            games_won += 1
            #end round one of the game
            return
        
        #give a hint    
        #if the guess is too high 
        elif guess > secret_number:
            print("Your guess is too high!")

    #if the guess is too low
        else:
            print("Your guess is too low!")

    #remove one attempt after an incorrect guess
        attempts -= 1

    #make the player aware how many attempts they have left
        print("Attempts remaining:", attempts)

    #if the loop finishes and the player has run out of attempts
    print("\n---------------------------------")
    print("       Game Over!       ")
    print("-----------------------------------")
    print("The number was:", secret_number)
```

```python
def display_statistics():
    #shows the player's games stats

    print("-----------STATISTICS-----------")
    print("Games played:", games_played)
    print("Games won:", games_won)
    print("Games lost:", games_lost)

    #calculate the player's win rate
    if games_played >0:
        win_rate = (games_won / games_played) * 100
        print("Win rate:", round(win_rate, 1), "%")

    print("------------------------------")
```

```python
#keep playing until the player decides to stop
while True:
    
    #time to play a new round of the game
    play_game()

    #keep asking until the player inputs a valid answer
    while True:
        #ask the player if they want to play again
        play_again = input("Would you like to play again? (yes/no): ")

        #check if the player wants to play again
        if play_again.lower() == "yes":
            break 

        #check if the player wants to stop playing
        elif play_again.lower() == "no":
            break

        #check if the player enters anything else
        else:
            print("Invalid input. Please enter yes or no.")
    #if the player says they want to stop the game
    if play_again.lower() == "no": 
        break
        
#shows the players stats
display_statistics()
```
